## Función `rango_edad`

Clasifica una edad en los rangos definidos en la sección 6 del EV02 (18-20, 21-23, 24-25). Las edades fuera de 18 a 25 se marcan como `fuera de rango`, porque no pertenecen a la población del estudio. `if` prueba la primera condición, `elif` solo se evalúa si la anterior fue falsa y `else` recoge el resto.

```python
def rango_edad(edad):
    """Clasifica una edad en los rangos del EV02 o la marca como fuera de rango."""
    if edad < 18 or edad > 25:
        return "fuera de rango"
    elif edad <= 20:
        return "18-20"
    elif edad <= 23:
        return "21-23"
    else:
        return "24-25"


for edad in [17, 18, 20, 21, 23, 24, 25, 26]:
    print(edad, "->", rango_edad(edad))
```
